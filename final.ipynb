```python
import requests
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

# FETCH DATA FROM API

url = "https://jsonplaceholder.typicode.com/users"
response = requests.get(url)
data = response.json()
print(data)
```

```python
users = [] # Create an empty list to store user data

for user in data:
    users.append({
        "Name": user["name"],
        "Email": user["email"],
        "City": user["address"]["city"],
        "Company": user["company"]["name"]
    })

df = pd.DataFrame(users)
print("\nUsers Data:\n")
print(df.to_string(index=False))
```

```python
scores = np.random.randint(50, 100, size=len(df))
df["Score"] = scores
print("\nUpdated Data:\n")
print(df)
```

```python
average_score = np.mean(scores)
print("\nAverage Score:", average_score)
city_count = df["City"].value_counts()
print("\nCity Count:\n")
print(city_count)
```

```python
plt.figure(figsize=(10, 5))
city_count.plot(kind='bar')
plt.title("Users Per City")
plt.xlabel("City")
plt.ylabel("Number of Users")
plt.show()
```
